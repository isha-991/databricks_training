```python
catalog_name = dbutils.widgets.get("catalog_name")
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog_name}")
spark.sql(f"USE CATALOG {catalog_name}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS bronze")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS silver")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS gold")
 
customers_staging_path = dbutils.widgets.get("customers_staging_path")
orders_staging_path = dbutils.widgets.get("orders_staging_path")
products_staging_path = dbutils.widgets.get("products_staging_path")
```

```python
customers_df = spark.read.json(path=customers_staging_path)
orders_df = spark.read.json(path=orders_staging_path)
products_df = spark.read.json(path=products_staging_path)
```

```python
customers_df.write.mode("append").format("DELTA").saveAsTable(f"bronze.customers")
```

```python
orders_df.write.mode("append").format("DELTA").saveAsTable(f"bronze.orders")
```

```python
products_df.write.mode("append").format("DELTA").saveAsTable(f"bronze.products")
```
