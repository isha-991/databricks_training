```python
from pyspark.sql.functions import *
```

```python
catalog_name = dbutils.widgets.get("catalog_name")
spark.sql(f"USE CATALOG {catalog_name}")
spark.read.table("bronze.orders").withColumn("total_amount",col("qty") * col("price")).write.mode("APPEND").saveAsTable("silver.orders")
```

```python
# spark.sql(f"USE CATALOG ecommerce")
order_df = spark.read.table("bronze.orders")
```

```python
order_df.columns
```

```python
customer_df = spark.read.table("bronze.customers")
```

```python
customer_df.columns
```

```python

joined_df = order_df.join(customer_df,on="customer_id",how="left")
```

```python
joined_df.columns
```

```python
product_df = spark.read.table("bronze.products")
```

```python
product_df.columns
```

```python
joined_df_ = joined_df.join(product_df,joined_df.item_id == product_df.product_id,how="left")
```

```python
joined_df_.columns
```

```python
display(joined_df_)
```

```python
joined_df_.drop('name').write.mode("APPEND").saveAsTable("silver.joined")
```
