## Custome Schema 

```python
from pyspark.sql.types import *
USER_SCHEMA = StructType([
    StructField("id", IntegerType()),
    StructField("age", IntegerType()),
    StructField("gender", StringType()),
    StructField("designation", StringType()),
    StructField("salary", IntegerType())
])
df = spark.read.csv(
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/sample_dataset/users_001.csv",header=True, schema=USER_SCHEMA )

display(df)
```

##handling bad data

```python
spark.read.json(path= "/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/access_logs.json").display()
```

```python
spark.read.option("columnNameOfCorruptRecord", "Bad_record").json(path= "/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/access_logs.json").display()
```

```python
spark.read.option("mode","FAILFAST").json(path= "/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/access_logs.json").display()
```

```python
spark.read.option("mode","DROPMALFORMED").json(path= "/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/access_logs.json").display()
```

```python
df = spark.read.csv(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/sample_dataset/users_001.csv", header=True, inferSchema=True)
df.write.save(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1",format="json")
spark.read.json(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1").display()
```

```python
df.write.save(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1",format="json",mode="overwrite")
spark.read.json(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1").count()
```

```python
df.write.save(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1",format="json",mode="append")
spark.read.json(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1").count()
```

```python
df.write.save(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1",format="json",mode="ignore")
spark.read.json(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/output/output_json1").count()
```
