##Loading Dataset

```python
employee_df = spark.read.csv(
    "/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/employee.csv",
    header=True,
    inferSchema=True,
    sep="|",
    quote="'",
).limit(8)
```

##Create a python Function

```python
def convert_to_title(name):
    return name.title()
```

##Applying function in pyspark

```python
employee_df.select(convert_to_title("name")).display()
```

```python
from pyspark.sql.functions import udf 
from pyspark.sql.types import StringType
convert_to_title_fn = udf(convert_to_title, StringType())
```

```python
employee_df.select(convert_to_title_fn("name").alias("name")).display()
```

```python
@udf(returnType=StringType())
def convert_to_title_udf(name):
  return name.title()
```

```python
employee_df.select(convert_to_title_udf("name").alias("name")).display()
```

```python
employee_df = spark.read.csv(
    "/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/employee.csv",
    header=True,
    inferSchema=True,
    sep="|",
    quote="'",
)
```

```python

from pyspark.sql.functions import col, when

employee_df.withColumn("name",when(col("name").isNotNull(), (convert_to_title_udf("name"))).otherwise(None)).select("name").display()
```

```python
from pyspark.sql.functions import when, col
 
employee_df.withColumn(
    "name",
    when(col("name").isNotNull(), convert_to_title_udf("name")).otherwise(col("name")),
).display()
 
```

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

convert_to_title_fn = udf(
    lambda e: convert_to_title(e) if e is not None else None,
    StringType()
)

employee_df.select(
    convert_to_title_fn("name").alias("name")
).display()
```
