```python
employee_df = spark.read.csv(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/employee_new.csv",header=True,inferSchema=True,quote='"')
employee_df.display()
```

##2.display first 10 rows

```python
display(employee_df.head(10))
```

##3.Print schema

```python
display(employee_df.printSchema())
```

##4.number of rows

```python
employee_df.count()
```

```python
employee_df.columns
```

## 6.Select only FirstName, LastName, and Salary.

```python
from pyspark.sql.functions import * 
display(employee_df.select(col('FirstName'),col('LastName'),col('Salary')))
```

##7.Show employees with salary greater than 60,000.

```python
display(employee_df.filter(col('Salary') > 60000))
```

## 8. Find employees who belong to DepartmentID = 1.

```python
display(employee_df.filter(col("DepartmentID") == 1).select(col('FirstName'),col('LastName')))
```

## 9. Filter employees hired after 2021-01-01.

```python
display(employee_df.filter(col("HireDate")>"2021-01-01").select(col('FirstName'),col('LastName')))
```

## 10. Display employees whose last name starts with "S".

```python
display(employee_df.filter(col("LastName").startswith('S')).select(col('FirstName'),col('LastName')))
```

## 11. Create a new column AnnualSalary = Salary * 12.

```python
employee_df = employee_df.withColumn("AnnualSalary",col("Salary")*12)
```

```python
display(employee_df.head(10))
```

## 12. Rename column PhoneNumber to Phone.

```python
employee_df=employee_df.withColumnRenamed("PhoneNumber","Phone")
display(employee_df.head(10))
```

## 13. Drop the column Email.

```python
employee_df = employee_df.drop(col("Email"))
```

```python
display(employee_df.head(10))
```

## 14. Cast Salary column to Integer.

```python
employee_df.printSchema()
```

```python
employee_df = employee_df.withColumn("Salary",col("Salary").cast("int"))
```

```python
employee_df.printSchema()
```

## 15. Add a constant column Country = "India".

```python
employee_df=employee_df.withColumn("Country", lit("India"))
```

```python
display(employee_df.head(10))
```

## 16. Find an average salary.

```python
display(employee_df.select(avg(employee_df.Salary).alias("Average Salary")))
```

## 17. Find maximum salary.

```python
display(employee_df.select(max(employee_df.Salary).alias("Maximum Salary")))
```

## 18. Count employees in each department.

```python
display(employee_df.groupBy(col("DepartmentName")).count())
```

## 19. Find total salary per department.

```python
display(employee_df.groupBy(col("DepartmentName")).agg(sum("Salary").alias("Total Salary")))
```

## 20. Find a department having the highest average salary.

```python

display(employee_df.groupBy("DepartmentName").agg(avg("Salary").alias("Total Salary")).orderBy(col("Total Salary").desc()).limit(1))
```

## 21. Sort employees by salary descending.

```python
employee_df = employee_df.orderBy(col("Salary").desc())
display(employee_df)
```

##22.Sort by HireDate ascending.

```python
employee_df = employee_df.orderBy(col("HireDate").asc())
display(employee_df)
```

##23.Find distinct DepartmentIDs.

```python

display(employee_df.select("DepartmentID").distinct())
```

##24.Show top 5 highest paid employees.

```python
display(employee_df.orderBy(col("Salary").desc()).limit(5))
```

##25.Remove duplicate employees (if any).

```python
employee_df = employee_df.dropDuplicates()
```
