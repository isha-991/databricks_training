##load 'user_001.csv' into dataframe

```python
df = spark.read.csv(path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/dataset/user_dataset/users_001.csv", header=True, inferSchema=True)
```

```python
df.display()
```

## Write as parquet 

```python
df.write.save(
    format="parquet",
    mode="overwrite",
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output/output_parquet",
)
 
```

## Reading parquet file

```python
df_parquet = spark.read.parquet(
    "/Volumes/quickstart_catalog/quickstart_schema/sandbox/output/output_parquet"
)
display(df_parquet)
 
```

##Creating a view

```python
df_parquet.createOrReplaceTempView("users_vw")
```

```python
%sql
UPDATE users_vw SET country = "Bharat" where country = "India"
```

## no support for roll back 

```python
from pyspark.sql.functions import col
df.filter(col("country")=="India").write.save(
    format="parquet",
    mode="overwrite",
    path="/Volumes/quickstart_catalog/quickstart_schema/sandbox/output/output_parquet",)
```

```python
df_parquet = spark.read.parquet(
    "/Volumes/quickstart_catalog/quickstart_schema/sandbox/output/output_parquet"
)
display(df_parquet)
 
```
